```python
class Human:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def check_adult(self):
        if self.age > 19:
            print(f"{self.name}は成人です。")
        else:
            print(f"{self.name}は未成年です。")

human_list = [["侍太郎", 28], ["侍二郎", 23], ["侍三郎", 19], ["侍花子", 30]]

human = []
for human_list_item in human_list:
    human.append(Human(human_list_item[0], human_list_item[1]))

for human_item in human:
    human_item.check_adult()
```
